```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N_list=np.linspace(0,70)
```

```python
def phi2_4(N):
    """
    Calculates phi_4 based on the number of e-folds N.
    
    Formula:
    7.11391 + 8 * N
    """
    return 7.11391 + 8.0 * N
```

```python
phi2_4_list = phi2_4(N_list)
plt.plot(N_list,phi2_4_list)
plt.xlim(max(N_list)+5, min(N_list)-5)
```

```python
def epsilon_4(phi2):
    """
    Calculates epsilon_4 taking phi^2 as the input argument.
    
    Formula:
    (0.16 / (0.04 * phi^2 + 1)) + (8 / (0.04 * (phi^2)^2 + phi^2))
    """
    # 1. Calculate Term 1
    # 0.16 / (0.04 * phi^2 + 1)
    denom1 = 0.04 * phi2 + 1.0
    term1 = 0.16 / denom1
    
    # 2. Calculate Term 2
    # 8 / (0.04 * (phi^2)^2 + phi^2)
    denom2 = (0.04 * phi2**2) + phi2
    term2 = 8.0 / denom2
    
    # 3. Final result
    return term1 + term2
```

```python
epsilon_4_list = abs(epsilon_4(phi2_4_list))
plt.plot(N_list,epsilon_4_list)
plt.xlim(max(N_list)+5, min(N_list)-5)
```

```python
def eta_4(phi2):
    """
    Calculates the updated eta parameter for the phi^4 potential.
    
    Parameters:
    phi2 (float or numpy.ndarray): The value of phi squared (phi^2).
    
    Returns:
    float or numpy.ndarray: The calculated value.
    """
    numerator = 0.64 * phi2 + 8.0
    denominator = 0.04 * (phi2**2) + phi2
    constant = 0.000150011
    
    return (numerator / denominator) + constant
```

```python
eta_4_list = abs(eta_4(phi2_4_list))
plt.plot(N_list,eta_4_list)
plt.xlim(max(N_list)+5, min(N_list)-5)
```

```python
def phi2_2_3(N):
    """
    Calculates phi^2_{2/3} based on the number of e-folds N.
    
    Formula:
    5.0 * (1 - exp((1/3) * (-0.8 * N - 0.127298)))
    """
    # 1. Calculate the term inside the exponential
    # (1/3) * (-0.8 * N - 0.127298)
    exponent = (1.0 / 3.0) * (-0.8 * N - 0.127298)
    
    # 2. Final result
    return 5.0 * (1.0 - np.exp(exponent))
```

```python
phi2_2_3_list = phi2_2_3(N_list)
plt.plot(N_list,phi2_2_3_list)
plt.xlim(max(N_list)+5, min(N_list)-5)
```

```python
def epsilon_2_3(phi2):
    """
    Calculates epsilon_{2/3} taking phi^2 as the input argument.
    
    Formula:
    (2 * (1 - 0.2 * phi^2)^2) / (9 * phi^2 * (0.04 * phi^2 + 1)) 
    - (0.0266667 * (0.2 * phi^2 - 1)) / (0.04 * phi^2 + 1)
    """
    # 1. Calculate Term 1
    # Numerator: 2 * (1 - 0.2 * phi^2)^2
    num1 = 2.0 * (1.0 - 0.2 * phi2)**2
    # Denominator: 9 * phi^2 * (0.04 * phi^2 + 1)
    den1 = 9.0 * phi2 * (0.04 * phi2 + 1.0)
    term1 = num1 / den1
    
    # 2. Calculate Term 2
    # Numerator: 0.0266667 * (0.2 * phi^2 - 1)
    num2 = 0.0266667 * (0.2 * phi2 - 1.0)
    # Denominator: 0.04 * phi^2 + 1
    den2 = 0.04 * phi2 + 1.0
    term2 = num2 / den2
    
    # 3. Final result
    return term1 - term2
```

```python
epsilon_2_3_list = abs(epsilon_2_3(phi2_2_3_list))
plt.plot(N_list,epsilon_2_3_list)
plt.xlim(max(N_list)+5, min(N_list)-5)
```

```python
def eta_2_3(phi2):
    """
    Calculates the eta parameter for the phi^(2/3) potential.
    
    Parameters:
    phi2 (float or numpy.ndarray): The value of phi squared (phi^2).
    
    Returns:
    float or numpy.ndarray: The calculated value.
    """
    numerator = 1.12 * phi2 + 4.0
    denominator = 0.12 * (phi2**2) + 3.0 * phi2
    constant = 0.000150011
    
    return (numerator / denominator) + constant
```

```python
eta_2_3_list = abs(eta_2_3(phi2_2_3_list))
plt.plot(N_list,eta_2_3_list)
plt.xlim(max(N_list)+5, min(N_list)-5)
```

```python
def phi2_1_3(N):
    """
    Calculates phi^2_{1/3} based on the number of e-folds N.
    
    Formula:
    2.27273 * (1 - exp((1/3) * (-0.88 * N - 0.0715332)))
    """
    # 1. Calculate the term inside the exponential
    # (1/3) * (-0.88 * N - 0.0715332)
    exponent = (1.0 / 3.0) * (-0.88 * N - 0.0715332)
    
    # 2. Final result
    return 2.27273 * (1.0 - np.exp(exponent))
```

```python
phi2_1_3_list = phi2_1_3(N_list)
plt.plot(N_list,phi2_1_3_list)
plt.xlim(max(N_list)+5, min(N_list)-5)
```

```python
def epsilon_1_3(phi2):
    """
    Calculates epsilon_{1/3} taking phi^2 as the input argument.
    
    Formula:
    ((1 - 0.44 * phi^2)^2) / (18 * phi^2 * (0.04 * phi^2 + 1)) 
    - (0.0133333 * (0.44 * phi^2 - 1)) / (0.04 * phi^2 + 1)
    """
    # 1. Calculate Term 1
    # Numerator: (1 - 0.44 * phi^2)^2
    num1 = (1.0 - 0.44 * phi2)**2
    # Denominator: 18 * phi^2 * (0.04 * phi^2 + 1)
    den1 = 18.0 * phi2 * (0.04 * phi2 + 1.0)
    term1 = num1 / den1
    
    # 2. Calculate Term 2
    # Numerator: 0.0133333 * (0.44 * phi^2 - 1)
    num2 = 0.0133333 * (0.44 * phi2 - 1.0)
    # Denominator: 0.04 * phi^2 + 1
    den2 = 0.04 * phi2 + 1.0
    term2 = num2 / den2
    
    # 3. Final result
    return term1 - term2
```

```python
epsilon_1_3_list = abs(epsilon_1_3(phi2_1_3_list))
plt.plot(N_list,epsilon_1_3_list)
plt.xlim(max(N_list)+5, min(N_list)-5)
```

```python
def eta_1_3(phi2):
    """
    Calculates the eta parameter for the phi^(1/3) potential.
    
    Parameters:
    phi2 (float or numpy.ndarray): The value of phi squared (phi^2).
    
    Returns:
    float or numpy.ndarray: The calculated value.
    """
    numerator = 1.04 * phi2 + 2.0
    denominator = 0.12 * (phi2**2) + 3.0 * phi2
    constant = 0.000150011
    
    return (numerator / denominator) + constant
```

```python
eta_1_3_list = abs(eta_1_3(phi2_1_3_list))
plt.plot(N_list,eta_1_3_list)
plt.xlim(max(N_list)+5, min(N_list)-5)
```

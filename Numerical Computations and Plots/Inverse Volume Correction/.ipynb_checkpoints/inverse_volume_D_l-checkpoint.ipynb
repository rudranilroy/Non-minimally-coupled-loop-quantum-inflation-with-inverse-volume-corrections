# Importing Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

# $n_s$-r calculation

```python
def D(l, q):
    sgn_q_minus_1 = np.sign(q-1)
    abs_q_minus_1 = abs(q - 1)
    
    term_1 = ((q + 1)**(l + 2) - abs_q_minus_1**(l + 2)) / (l + 2)
    term_2 = (q * ((q + 1)**(l + 1) - (sgn_q_minus_1 * abs_q_minus_1**(l + 1)))) / (l + 1)
    
    main_numerator = 3 * (q**(1 - l)) * (term_1 - term_2)
    
    base = main_numerator / (2 * l)
    exponent = 3 / (2 - 2 * l)
    
    return base ** exponent
```

```python
ql=np.linspace(0,150)
dl = D(3/4,ql)
horizony = np.linspace(0,1.4)
horizonx = 100 + (np.linspace(0,1.5)*0)
plt.plot(ql,dl,color="blue")
plt.plot(horizonx,horizony,color="red",label="Horizon Crossing")
plt.xlabel("$q = a_{LQ}^2/a_{Pl}^2$",fontsize=15)
plt.ylabel("$D_{3/4}(q)$",fontsize=15)
plt.legend()
plt.savefig("D_l.pdf", dpi=300)
plt.show()
```

```python
ql_zoom=np.linspace(0,3,10000)
dl_zoom = D(3/4,ql_zoom)
plt.plot(ql_zoom,dl_zoom,color="blue")
plt.plot(ql_zoom,1*(dl_zoom**0),color="red", linestyle='dashed')
plt.xlabel("$q = a_{LQ}^2/a_{Pl}^2$",fontsize=15)
plt.ylabel("$D_{3/4}(q)$",fontsize=15)
plt.xlim(0,3)
plt.savefig("D_l_zoom.pdf", dpi=300)
plt.show()
```
